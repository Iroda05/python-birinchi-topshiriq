```python
import matplotlib.pyplot as plt

kunlar = ["Du", "Se", "Chor", "Pay", "Ju", "Shan", "Yak"]
uyqu = [7, 6, 8, 5, 7, 9, 10]

plt.bar(kunlar, uyqu, color="teal")
plt.axhline(8, color="red", linestyle="--", label="Tavsiya: 8 soat")

plt.title("Mening haftalik uyqum")
plt.ylabel("Soat")
plt.legend()
plt.show()

print("O‘rtacha uyqu:", round(sum(uyqu) / len(uyqu), 1), "soat")
```
